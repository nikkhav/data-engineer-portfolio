```python
import os
from dotenv import load_dotenv
import boto3
import pandas as pd
from io import BytesIO
```

```python
load_dotenv()
```

```python
S3_KEY_ID = os.getenv("S3_KEY_ID")
S3_SECRET = os.getenv("S3_SECRET")
S3_ENDPOINT = os.getenv("S3_ENDPOINT")
S3_BUCKET = os.getenv("S3_BUCKET")
```

```python
s3 = boto3.client(
    "s3",
    endpoint_url=S3_ENDPOINT,
    aws_access_key_id=S3_KEY_ID,
    aws_secret_access_key=S3_SECRET
)
```

```python
response = s3.list_objects_v2(Bucket=S3_BUCKET)

files = [obj["Key"] for obj in response.get("Contents", [])]
files
```

```python
currencies_history = "currencies_history.csv"
transactions_batch = "transactions_batch_1.csv"

currencies = pd.read_csv(BytesIO(s3.get_object(Bucket=S3_BUCKET, Key=currencies_history)["Body"].read()))
transactions = pd.read_csv(BytesIO(s3.get_object(Bucket=S3_BUCKET, Key=transactions_batch)["Body"].read()))
```
